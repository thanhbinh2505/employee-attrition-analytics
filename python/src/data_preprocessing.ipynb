```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import LabelEncoder, MinMaxScaler
```

```python
# Loading the dataset to df variable
df = pd.read_csv('../../Data/raw_data/HR-Employee-Attrition.csv')
print("Dataset import successful!")
```

```python
# View dataset
df.head()
```

```python
# Set the max rows and columns to display
pd.options.display.max_rows = len(df)
pd.options.display.max_columns = len(df.columns)
```

```python
df.describe(include='all')
```

```python
df.drop(columns=["EmployeeCount", "StandardHours", "Over18", "DailyRate", "HourlyRate", "MonthlyRate"], inplace=True) 
```

```python
df.head(5)
```

```python
binary_cols = ['Gender', 'OverTime', 'Attrition']
le  = LabelEncoder()
for col in binary_cols:
    # Chuyển đổi: YES/NO -> 1/0, MALE/FEMALE -> 1/0
    df[col] = le.fit_transform(df[col])
```

```python
nominal_cols = ['Department', 'EducationField', 'JobRole', 'MaritalStatus']
df = pd.get_dummies(df, columns=nominal_cols, dtype=int)
df.head(5)
```

```python
scaler = MinMaxScaler()
scaler_cols = [
    'Age', 'DistanceFromHome','MonthlyIncome', 'NumCompaniesWorked', 
    'PercentSalaryHike', 'TotalWorkingYears', 'TrainingTimesLastYear', 
    'YearsAtCompany', 'YearsInCurrentRole', 'YearsSinceLastPromotion', 
    'YearsWithCurrManager'
]
df[scaler_cols] = scaler.fit_transform(df[scaler_cols])
df.head(5)
```

```python
scaler1_cols = [
    'Education', 'EnvironmentSatisfaction', 'JobInvolvement', 
    'JobLevel', 'JobSatisfaction', 'PerformanceRating', 
    'RelationshipSatisfaction', 'StockOptionLevel', 'WorkLifeBalance'
]
df[scaler1_cols] = scaler.fit_transform(df[scaler1_cols])
df.head(5)
```

```python
#Chuyển thành dạng 0 1 2
travel_map = {
    'Non-Travel': 0,
    'Travel_Rarely': 1,
    'Travel_Frequently': 2
}
df['BusinessTravel'] = df['BusinessTravel'].map(travel_map)
df.head(5)
```

```python
#Chuyển về khoảng cách 0-1
df[['BusinessTravel']] = scaler.fit_transform(df[['BusinessTravel']])
df.head(20)
```

```python
# Lưu dữ liệu đã tiền xử lý
df.to_csv('../../Data/processed_data/HR-Employee-Attrition-Processed.csv', index=False)

print("Lưu thành công")
```
